Create and Read a CSV File

```python
import csv

with open("sales.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Product", "Sales"])
    writer.writerow(["Laptop", 50000])
    writer.writerow(["Mobile", 25000])
    writer.writerow(["Tablet", 15000])

with open("sales.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
try:
    number = int(input("Enter a number: "))
    print("Your number is:", number)

except ValueError:
    print("Invalid input! Please enter a valid number.")

finally:
    print("Program completed.")
```

Read a CSV File, Check Whether It Exists, and Handle Invalid Data in

```python
import csv
import os

filename = "sales.csv"

try:
    if not os.path.exists(filename):
        print("File does not exist")

    else:
        with open(filename, "r") as file:
            reader = csv.DictReader(file)

            for row in reader:
                try:
                    sales = float(row["Sales"])
                    print(row["Product"], "Sales:", sales)

                except (ValueError, KeyError):
                    print("Invalid data:", row)

except (OSError, csv.Error) as error:
    print("Error:", error)

finally:
    print("Program completed")
```

Read the CSV file and handle errors

```python
import csv
import os

try:
    if os.path.exists("sales.csv"):
        with open("sales.csv", "r") as file:
            reader = csv.DictReader(file)

            for row in reader:
                try:
                    sales = float(row["Sales"])
                    print(row["Product"], sales)
                except ValueError:
                    print("Invalid sales data")

    else:
        print("File does not exist")

except OSError as error:
    print("Error:", error)

finally:
    print("Program completed")
```

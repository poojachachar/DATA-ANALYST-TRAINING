```python
import matplotlib.pyplot as plt

products = ['laptop','mobile','tablet','monitor']
sales = [50000,80000,30000,40000]

plt.bar(products, sales)
plt.title('product-wise sales')
plt.xlabel('products')
plt.ylabel('sales amount')
plt.show()
```

Business Observation: Mobile phones have the highest sales (₹80,000), while tablets have the lowest sales (₹30,000).

line chart

```python
import matplotlib.pyplot as plt

months = ['jan','feb','mar','apr','may']
sales = [20000,25000,22000,30000,35000]

plt.plot(months,sales,marker='o',label='monthly sales')
plt.title('monthly sales trend')
plt.xlabel('month')
plt.ylabel('sales amount')
plt.legend()
plt.show()
```

Business Observation: Sales show an overall increasing trend from January to May, indicating business growth

HISTOGRAM

```python
import matplotlib.pyplot as plt

salary = [20000, 25000, 25000, 30000, 35000,
          40000, 45000, 50000, 55000, 60000]

plt.hist(salary, bins=5,edgecolor='black')
plt.title('employee salary distribution')
plt.xlabel(salary)
plt.ylabel('number of employees')
plt.show()
```

Business Observation: The histogram helps HR understand which salary ranges contain the most employees.

BOX PLOT

```python
import matplotlib.pyplot as plt

salary = [20000, 25000, 25000, 30000, 35000,
          40000, 45000, 50000, 55000, 150000]

plt.boxplot(salary)
plt.title('Employee Salary Box Plot')
plt.ylabel('Salary')
plt.show()
```

Business Observation: The box plot helps HR identify unusually high or low salaries that may need further investigation.

SCATTER PLOT

```python
import matplotlib.pyplot as plt

advertising = [1000, 2000, 3000, 4000, 5000]
sales = [10000, 18000, 25000, 35000, 45000]

plt.scatter(advertising, sales)
plt.title('Advertising Expenses vs Sales')
plt.xlabel('Advertising Expenses (₹)')
plt.ylabel('Sales (₹)')
plt.show()
```

Business Observation: The scatter plot shows a positive relationship between advertising expenses and sales. As advertising expenses increase, sales also tend to increase.

CORRELATION HEATMAP

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Sales': [100, 200, 300, 400, 500],
    'Advertising': [10, 20, 30, 40, 50],
    'Profit': [20, 30, 50, 60, 80]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Business Observation: Sales and advertising have a strong positive correlation in this sample dataset. This means they tend to increase together, but it does not prove that advertising causes higher sales.

```python
import matplotlib.pyplot as plt

months = ['jan','feb','mar','apr','may']
sales = [20000,25000,22000,30000,35000]

plt.plot(months,sales,marker='o',label='monthly sales')

plt.title('monthly sales analysis')
plt.xlabel('month')
plt.ylabel('sales amount (rupee)')
plt.legend()
plt.grid(True)
plt.show()
```

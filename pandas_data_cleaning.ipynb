identify and handle missing value


```python
import pandas as pd

data = {




    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Pooja", "Rahul", "Amit", "Neha", "Priya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, None, 40000, None, 50000],
    "Age": [25, 30, None, 28, 35]
}

df = pd.DataFrame(data)

print(df)
```

identify missing value

```python
print(df.isnull().sum())
```

handle missing values

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

verify missing value

```python


print(df.isnull().sum())
```

Convert Date Columns into Proper Datetime Values

Add a date column to your dataset

```python
df["Joining_Date"] = [
    "2024-01-10",
    "2024-02-15",
    "2024-03-20",
    "2024-04-25",
    "2024-05-30"
]

print(df[["Name", "Joining_Date"]])
```

Convert the column into datetime

```python

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df["Joining_Date"].dtype)
```

Display the converted dates

```python
print(df[["Name", "Joining_Date"]])
```

ADD into time

```python
df["Joining_Date"] = [
    "2024-01-10 09:30:00",
    "2024-02-15 10:15:00",
    "2024-03-20 11:00:00",
    "2024-04-25 12:30:00",
    "2024-05-30 14:45:00"
]

print(df[["Name", "Joining_Date"]])
```

```python

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"]).dt.date

df["Joining_Time"] = [
    "09:30:00",
    "10:15:00",
    "11:00:00",
    "12:30:00",
    "14:45:00"
]

print(df[["Name", "Joining_Date", "Joining_Time"]])
```

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics',
                 'Electronics', 'Electronics', 'Electronics'],
    'Sales': [50000, 30000, 20000, 60000, 25000, 15000],
    'Profit': [8000, 5000, 2000, 10000, -1000, 1500],
    'Quantity': [5, 10, 8, 6, 12, 7],
    'City': ['Pune', 'Mumbai', 'Pune', 'Mumbai', 'Pune', 'Mumbai'],
    'Date': ['2026-01-10', '2026-01-15', '2026-02-05',
             '2026-02-12', '2026-02-20', '2026-03-01']
}

df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])

df
```

Find the product with the highest sales

```python
df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
```

Find the city with the highest sales

```python
df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
```

find monthly sales

```python
df.groupby(df['Date'].dt.month_name())['Sales'].sum()
```

Find products with low or negative profit

```python
df.groupby('Product')['Profit'].sum().sort_values()
```

visualization

```python
import matplotlib.pyplot as plt

product_sales = df.groupby('Product')['Sales'].sum()

product_sales.plot(kind='bar')

plt.title('Product-wise Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')

plt.show()
```

City-wise Sales Bar Chart

```python
city_sales = df.groupby('City')['Sales'].sum()

city_sales.plot(kind='bar')

plt.title('City-wise Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')

plt.show()
```

Monthly Sales Trend

```python
monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.show()
```

Product-wise Profit

```python
product_profit = df.groupby('Product')['Profit'].sum()

product_profit.plot(kind='bar')

plt.title('Product-wise Total Profit')
plt.xlabel('Product')
plt.ylabel('Total Profit')

plt.show()
```

Category-wise Profit

```python
category_profit = df.groupby('Category')['Profit'].sum()

category_profit.plot(kind='bar')

plt.title('Category-wise Total Profit')
plt.xlabel('Category')
plt.ylabel('Total Profit')

plt.show()
```

```python
import numpy as np
```

```python
print(np.array([10,20,30]))
```

Create One-Dimensional and Multidimensional Arrays

```python
arr=np.array([10,20,30,40,50])
print(arr)
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
```

Indexing, Slicing, Filtering and Reshaping

Indexing,

```python
arr=np.array([10,20,30,40,50])
print(arr[0])
```

Slicing:

```python
arr=np.array([10,20,30,40,50])
print(arr[1:4])
```

Filtering:

```python
arr=np.array([10,20,30,40,50])
print(arr[arr>25])
```

reshaping


```python
arr=np.array([1,2,3,4,5,6])
print(arr.reshape(2, 3))
```

NUMERICAL CALCULATIONS

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.sum(arr))
```

MAX

```python
arr=np.array([10,20,30,40,50])
print(np.max(arr))
```

MIN

```python
arr=np.array([10,20,30,40,50])
print(np.min(arr))
```

MEAN

```python
arr=np.array([10,20,30,40,50])
print(np.mean(arr))
```

standard deviation

```python
arr=np.array([10,20,30,40,50])
print(np.std(arr))
```

Python vs NumPy Calculations

```python
a = [10, 20, 30]
print([x + 5 for x in a])
```
